```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Исходная матрица:\n", arr)

print("Главная диагональ (индексами):", arr[0, 0], arr[1, 1], arr[2, 2])
print("Главная диагональ (np.diag):", np.diag(arr))

print("Побочная диагональ (индексами):", arr[0, 2], arr[1, 1], arr[2, 0])
print("Побочная диагональ (np.diag):", np.diag(np.fliplr(arr)))
print("Побочная диагональ (срезы):", arr[np.arange(arr.shape[0]), np.arange(arr.shape[1]-1, -1, -1)])
```

```python
import pandas as pd

data = {
    'Имя': ['Аня', 'Борис', 'Вика', 'Гриша'],
    'Возраст': [23, 35, 22, 45],
    'Город': ['Москва', 'Санкт-Петербург', 'Казань', 'Новосибирск']
}
df = pd.DataFrame(data)
print("Исходный DataFrame:\n", df)

df['Зарплата'] = [95000, 140000, 80000, 160000]

df['Премия'] = df['Зарплата'] * 0.10

df['Стаж_лет'] = [1.5, 8.0, 0.5, 15.0]

print("\nDataFrame после добавления новых столбцов:\n", df)

new_employees = pd.DataFrame([
    {'Имя': 'Дмитрий', 'Возраст': 29, 'Город': 'Москва', 'Зарплата': 120000, 'Премия': 12000, 'Стаж_лет': 4.0},
    {'Имя': 'Елена', 'Возраст': 33, 'Город': 'Казань', 'Зарплата': 110000, 'Премия': 11000, 'Стаж_лет': 5.5}
])

df = pd.concat([df, new_employees], ignore_index=True)
print("\nDataFrame после добавления строк:\n", df)

# Сотрудники с зарплатой более 100 000 руб.
filter_salary = df[df['Зарплата'] > 100000]
print("\nСотрудники с зарплатой > 100 000:\n", filter_salary)

# Сотрудники старше 25 лет из Москвы или Казани
complex_filter = df[(df['Возраст'] > 25) & (df['Город'].isin(['Москва', 'Казань']))]
print("\nСотрудники старше 25 лет из Москвы или Казани:\n", complex_filter)
```

```python
import numpy as np


def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1.0 - s)

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
y = np.array([[0], [1], [1], [0]])

np.random.seed(42)
weights = np.random.randn(2, 1)
bias = 0.0
lr = 0.1

for epoch in range(10000):
    z = np.dot(X, weights) + bias
    y_pred = sigmoid(z)
    error = y_pred - y
    
    dcost_dz = error * sigmoid_derivative(z)
    weights -= lr * np.dot(X.T, dcost_dz)
    bias -= lr * np.sum(dcost_dz)

print("--- Результат однослойного перцептрона для XOR ---")
print("Предсказания (около 0.5 для всех, не разделяет):")
print(np.round(y_pred, 3))


np.random.seed(42)

W1 = np.random.randn(2, 2)
b1 = np.zeros((1, 2))
W2 = np.random.randn(2, 1)
b2 = np.zeros((1, 1))

lr = 0.5
epochs = 10000

for epoch in range(epochs):
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    
    z2 = np.dot(a1, W2) + b2
    y_pred = sigmoid(z2)
    
    error = y_pred - y
    delta2 = error * sigmoid_derivative(z2)
    
    delta1 = np.dot(delta2, W2.T) * sigmoid_derivative(z1)
    
    W2 -= lr * np.dot(a1.T, delta2)
    b2 -= lr * np.sum(delta2, axis=0, keepdims=True)
    W1 -= lr * np.dot(X.T, delta1)
    b1 -= lr * np.sum(delta1, axis=0, keepdims=True)

print("\n--- Результат перцептрона для XOR ---")
for i in range(len(X)):
    print(f"Вход: {X[i]} -> Предсказание: {y_pred[i][0]:.4f} (Истина: {y[i][0]})")
```
